```python
import random
import os
import re

while(1<2):
    print("\n");
    print("Let's Begin: Rock-Scissor-Paper: Shoot.....")

    userinput = input("Select your choice: [R]ock [P]aper [S]cissor [E]xit:");
    #print(len(userinput));

    if(not re.match("[SsRrPpEe]",userinput) or (len(userinput)) != 1):
        print("\n Enter right choice, buddy \n");
        continue;
        
    elif(re.match("[Ee]",userinput)):
        print("Ending Game !");
        break;
        
    print("You chose:"+userinput)
    choice = ['R','P','S']
               
    opp_turn = random.choice(choice)
    print("Computer chose:"+opp_turn)

    if(opp_turn == str.upper(userinput)):
        print("Tie")
        continue;
    elif(userinput.upper() == 'R' and opp_turn == 'P'):
        print("Paper catch Rock, I Win")
        continue;
    elif(userinput.upper() == 'S' and opp_turn == 'R'):
        print("Rock breaks Scissor, I Win")
        continue;
    elif(userinput.upper() == 'P' and opp_turn == 'S'):
        print("Scissors cuts Paper, I Win")
        continue;
    else:
        print("YOU WIN !!")
    
```
